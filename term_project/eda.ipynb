# Air-quality exploratory data analysis

This notebook runs and displays the reproducible EDA in `eda_analysis.py`. It covers distributions, correlations, pollutant relationships, and time/calendar patterns for both datasets. All analytical plots use the existing **training sample only**; the test exports are not loaded.

The generated report and CSV tables are available in `eda/`. Existing preprocessing and model input files are preserved. Run the cells in order, with the notebook opened from the repository root or `term_project/`.

```python
from pathlib import Path
import runpy
import pandas as pd
from IPython.display import display, Image, Markdown

project_dir = Path.cwd()
if not (project_dir / "eda_analysis.py").exists():
    project_dir = project_dir / "term_project"
assert (project_dir / "eda_analysis.py").exists(), "Open from the repository root or term_project directory."
project_dir = project_dir.resolve()
runpy.run_path(str(project_dir / "eda_analysis.py"), run_name="__main__")
eda_dir = project_dir / "eda"
```

## Findings and interpretation

The report below is generated from the data on each run. Its figures are displayed in the later cells.

```python
report = (eda_dir / "EDA_REPORT.md").read_text()
# Images are displayed explicitly below so they work from either working directory.
text_only = "\n".join(line for line in report.splitlines() if not line.startswith("![") and not line.startswith("### "))
display(Markdown(text_only))
```

## Dataset accounting

Full-source counts describe preprocessing. Analytical statistics and figures use only the training rows.

```python
display(pd.read_csv(eda_dir / "tables/dataset_overview.csv"))
```

## Pollutant distributions

Histograms include every retained training observation. The count axis is logarithmic so sparse tails remain visible. Extreme concentrations are not automatically errors.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_distributions.png")))
```

## PM2.5 by location

Compare medians and spreads across locations. Outlier dots are hidden only in these boxplots; no observations are removed.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_location_distributions.png")))
```

## Correlations between current measurements

Pearson correlations show linear association. The tables also include Spearman and within-location-centered correlations to help assess sensitivity to skew and location differences.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_correlations.png")))
```

## Current features and future PM2.5

Each column is a separate forecast horizon. These marginal training correlations are exploratory evidence, not feature selection or forecast accuracy.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_forecast_correlations.png")))
```

## Relationships between pollutants

Hexagonal bins show density without hiding large numbers of overlapping observations. The axes use source values; color represents observation counts on a logarithmic scale.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_pollutant_relationships.png")))
```

## Daily patterns and coverage

Daily location means show temporal variation; row counts show changing coverage. Missing full days appear as gaps.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_time_patterns.png")))
```

## Monthly and hourly patterns

Global training covers only January–September and its hours are UTC. Beijing month profiles pool multiple training years. Month-of-year patterns should not be treated as shared climate seasons across countries or proof of causation.

```python
for dataset in ["global", "beijing"]:
    display(Image(filename=str(eda_dir / "figures" / f"{dataset}_calendar_patterns.png")))
```

## Numerical tables

Use these statistics in the presentation/report. The remaining correlations, location summaries, calendar counts, and aligned training metadata are in `eda/tables/`.

```python
for dataset in ["global", "beijing"]:
    display(Markdown(f"### {dataset.title()}"))
    display(pd.read_csv(eda_dir / "tables" / f"{dataset}_descriptive_statistics.csv", index_col=0).round(3))
    display(pd.read_csv(eda_dir / "tables" / f"{dataset}_forecast_correlations.csv", index_col=0).round(3))
```
